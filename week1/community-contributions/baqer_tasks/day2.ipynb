```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

def fetch_article_text(url):
    driver = webdriver.Chrome()
    driver.get(url)

    wait = WebDriverWait(driver, 10)

    element = wait.until(
        EC.presence_of_element_located((
            By.XPATH,
            '//*[@id="article-0"]/div[2]/div[5]'
        ))
    )
    text = element.text
    driver.quit()
    return text
```

```python

import requests
from bs4 import BeautifulSoup
from IPython.display import Markdown, display
import json
```

```python
OLLAMA_API = "http://localhost:11434/api/chat"
HEADERS = {"Content-Type": "application/json"}
MODEL = "llama3.2:1b"
```

```python
messages = [
    {"role": "system", "content": "Summarize the article in a snarky tone using Markdown in arabic ."},
    {"role": "user", "content": fetch_article_text("https://www.alaraby.co.uk/politics/%D9%85%D8%A8%D8%B9%D9%88%D8%AB-%D8%AA%D8%B1%D8%A7%D9%85%D8%A8-%D8%A5%D9%84%D9%89-%D8%A7%D9%84%D8%B9%D8%B1%D8%A7%D9%82-%D9%8A%D8%B1%D8%AD%D9%91%D8%A8-%D8%A8%D9%85%D9%88%D8%A7%D9%81%D9%82%D8%A9-%D9%81%D8%B5%D8%A7%D8%A6%D9%84-%D8%B9%D9%84%D9%89-%D9%86%D8%B2%D8%B9-%D8%A7%D9%84%D8%B3%D9%84%D8%A7%D8%AD") }
    
]
```

```python
payload = {
        "model": MODEL,
        "messages": messages,
        "stream": True
    }
```

```python
with requests.post(OLLAMA_API, json=payload, headers=HEADERS ,stream=True) as response:
    for line in response.iter_lines():
        if line:
            data = json.loads(line.decode("utf-8"))
            print(data["message"]["content"], end='', flush=True)
```
